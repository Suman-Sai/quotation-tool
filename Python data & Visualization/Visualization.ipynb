```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('ggplot')
# getting the data from the txt file and convert it intothe python...!
records=pd.read_csv(r"C:\Users\suman\Downloads\insurance_data_with_day.csv")
records
```

```python
insurance_groupby_day=records.groupby('dayofmonth')
days=insurance_groupby_day['dayofmonth'].mean()
plt.plot(days,insurance_groupby_day['age'].mean(),marker = 'o',label='Age')
plt.plot(days,insurance_groupby_day['bmi'].mean(),marker = 'o',label='BMI')
plt.plot(days,insurance_groupby_day['bloodpressure'].mean(),marker = 'o',label='Blood Pressure')
plt.xlabel('days')
plt.ylabel('Measure')
plt.title('Hospital Data')
plt.legend(loc='center left')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data={"Sales":[200,450,800,1200,650,950,300], "Profit":[40,80,100,150,60,120,50], "Discount":[0.05,0.10,0.12,0.15,0.20,0.25,-.30],
      "Category":["Furniture","Technology","Office Supplies","Furniture","Technology","Office Supplies","Furniture"]}

df=pd.DataFrame(data)
df
```

```python
# Advanced Scatter Plot
sns.scatterplot(data=df, x='Sales', y='Profit', 
                hue='Category')
plt.title("Sales vs Profit by Category and Discount")
plt.show()
```

```python
sns.catplot(x='Category', y='Sales', data=df, kind='bar')
plt.show()
sns.catplot(x='Category', y='Sales', data=df, kind='box')
plt.show()
```

```python
sns.catplot(x='Category', y='Sales', data=df, kind='violin')
plt.show()
```

```python
# Sample data
#categories = df['Category']
#sales = df['Sales']
#quantity = df['Profit']
category_data = df.groupby('Category').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
categories = category_data['Category']
sales = category_data['Sales']
quantity =category_data['Profit']
    
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Top-left: Bar plot
axes[0,0].bar(categories, sales, color="skyblue")
axes[0,0].set_title("Sales by Category")

# Top-right: Line plot
axes[0,1].plot(categories, sales, marker="o")
axes[0,1].set_title("Sales Trend")

# Bottom-left: Scatter plot
axes[1,0].scatter(sales, quantity, c="red")
axes[1,0].set_title("Sales vs Quantity")

# Bottom-right: Pie chart
axes[1,1].pie(sales, labels=categories, autopct='%1.1f%%')
axes[1,1].set_title("Sales Share")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt # Added this line

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, axes = plt.subplots(2, 1, sharex=True, figsize=(8, 6))

axes[0].plot(x, y1, color="blue")
axes[0].set_title("Sine Function")

axes[1].plot(x, y2, color="green")
axes[1].set_title("Cosine Function")

plt.tight_layout()
plt.show()
```
