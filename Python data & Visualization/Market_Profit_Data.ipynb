```python
test1=test1.assign(ProfitorLoss=np.where(test1['Sales']>35000,'Profit','Loss'))
test1.drop('Profit', axis=1,inplace=True)
```

```python
#Which Month getting profits list...!
print('The Total Profits Count is: ',(test1['ProfitorLoss']=='Profit').sum())
test1.loc[test1['ProfitorLoss'] == 'Profit',['Month','Branch','Sales','ProfitorLoss']]
```

```python
##Which Month getting Losses list...!
print('The Total Profits Count is: ',(test1['ProfitorLoss']=='Loss').sum())
test1.loc[test1['ProfitorLoss'] == 'Loss',['Month','Branch','Sales','ProfitorLoss']]
```
