###### Q. A company offers its customers to take a vehicle and an educational loan. A study shows that the probability that a customer takes a vehicle loan is 0.35, the probability that a customer takes an educational loan is 0.46 and the probability that a customer takes at least one of the loan is 0.59. Find the probability that a customer takes both vehicle and educational loans.

```python
# Given probabilities
p_vehicle = 0.35          # P(A)
p_educational = 0.46      # P(B)
p_at_least_one = 0.59     # P(A or B)

# Calculating the probability of taking both loans P(A and B)
# Formula: P(A and B) = P(A) + P(B) - P(A or B)
p_both = p_vehicle + p_educational - p_at_least_one

print(f"The probability that a customer takes both loans is: {p_both:.2f}")
```

###### Q. Consider a vending machine containing 6 boxes each of four different juice flavors (Mango, Apple, Guava and Kiwi). Find the probability that the juice box picked is either of Mango or Guava flavor.

```python
# Define the total number of boxes for each flavor
total_boxes_per_flavor = 6

# The machine contains 4 different flavors
flavors = ['Mango', 'Apple', 'Guava', 'Kiwi']
num_flavors = len(flavors)

# Calculate total number of juice boxes in the vending machine
total_juice_boxes = total_boxes_per_flavor * num_flavors

# Number of favorable outcomes (boxes that are either Mango or Guava)
favorable_boxes = total_boxes_per_flavor + total_boxes_per_flavor

# Calculate the probability
probability = favorable_boxes / total_juice_boxes

# Print the results
print(f"Total juice boxes in the machine: {total_juice_boxes}")
print(f"Number of Mango or Guava boxes: {favorable_boxes}")
print(f"Probability of picking Mango or Guava: {probability:.2f} (or {favorable_boxes}/{total_juice_boxes})")
```

###### Q. Consider an random variable that follows normal distribution with mean 5 and variance 10
###### Obtain: P(X < 1) & P(x > 5) & P(3 < x < 7)

```python
import numpy as np
import scipy.stats as stats

# Given parameters
mean = 5
variance = 10
std_dev = np.sqrt(variance) # Standard deviation is sqrt(variance)

# Creating the normal distribution object to store the parameters
norm_dist = stats.norm(loc=mean, scale=std_dev)

# 1. Obtain P(X < 1)
prob_less_than_1 = norm_dist.cdf(1)

# 2. Obtain P(X > 5): it can be any number like 6, 7,8,9,10,--------.
prob_greater_than_5 = 1 - norm_dist.cdf(5)

# 3. Obtain P(3 < X < 7)
prob_between_3_and_7 = norm_dist.cdf(7) - norm_dist.cdf(3)

# Print results
print(f"Normal Distribution: {norm_dist}")
print(f"1. P(X < 1)      = {prob_less_than_1:.4f}")
print(f"2. P(X > 5)      = {prob_greater_than_5:.4f}")
print(f"3. P(3 < X < 7)  = {prob_between_3_and_7:.4f}")
```
