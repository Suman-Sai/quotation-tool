```python
import matplotlib.pyplot as plt
import numpy as np

# Line - Single Line, Multiline, Bar - Verical, Horizontal, Grouped, Stacked, Boxplot, Hitogram, Pie Plot, Scatter
plt.style.use("ggplot")

month=np.arange(1,13)
prices=[101,112,203,304,335,406,507,608,709,810,795,854]

plt.plot(month, prices, color="r", marker="o")
plt.xlabel("Month")
plt.ylabel("Price")
plt.title("Monthly Price Trend")
plt.show()
```

```python
import pandas as pd

df_insurance=pd.read_csv(r"D:\Backup 14.03.2021\Desktop\Imarticus Post\Online PGA 38\Python Prog\Datasets\insurance_data_with_day.csv")
df_insurance
```

```python
df_ins_groupby_day=df_insurance.groupby("dayofmonth")
days=df_ins_groupby_day["dayofmonth"].mean()

plt.plot(days,df_ins_groupby_day["age"].mean(), color="limegreen", marker=".", label="Age")
plt.plot(days,df_ins_groupby_day["bmi"].mean(), color="crimson", marker=".", label="BMI")
plt.plot(days,df_ins_groupby_day["bloodpressure"].mean(), color="goldenrod", marker=".", label="Blood Pressure")

plt.title("Multi-Line Plots")
plt.xlabel("Days")
plt.ylabel("Measure")
plt.legend(loc="center right")
plt.show()
```

```python
df_ins_groupby_day_gender=df_insurance.groupby(["dayofmonth","gender"])
df_ins_groupby_day_gender["claim"].mean().unstack()
```

```python
fig,ax=plt.subplots(figsize=(7,4))

df_ins_groupby_day_gender=df_insurance.groupby(["dayofmonth","gender"])
df_ins_groupby_day_gender["claim"].mean().unstack().plot(ax=ax)
plt.show()
```

```python
fig, ax=plt.subplots(2,2)

x=np.linspace(0,8,1000)

ax[0,0].plot(x, x**2, "g")
ax[0,0].title.set_text("Axis [0,0]")

ax[1,0].plot(x, x**3, "k")
ax[1,0].title.set_text("Axis [1,0]")

ax[0,1].plot(x, np.sin(x), "b")
ax[0,1].title.set_text("Axis [0,1]")

ax[1,1].plot(x, np.cos(x), "r")
ax[1,1].title.set_text("Axis [1,1]")

fig.show()
```
