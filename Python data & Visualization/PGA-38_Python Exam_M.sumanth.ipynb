### Q1. The E-Commerce Discount Engine [Easy] (3 Marks)
**Scenario:**
Write a function `calculate_final_price(price, is_member, coupon_code)`:
1. Members (`is_member=True`) get a flat 10% discount.
2. If `coupon_code` is "FESTIVE20", take an *additional* $20 off (after member discount).
3. Price cannot be negative.


````python
def calculate_final_price(price, is_member, coupon_code):
    # 1. Members (is_member=True) get a flat 10% discount
    if is_member:
        price *= 0.90
    # 2.If coupon_code is "FESTIVE20", take an additional $20 off (after member discount).
    if coupon_code == "FESTIVE20":
        price -= 20
    # 3. Price cannot be negative
    return max(0, price)
#result 
print('The Overall Discount is:',calculate_final_price(150,False,"FESTIVE20"))
````

### Q2. Parking Lot Fee Calculator [Medium] (4 Marks)
**Scenario:**
Calculate fee based on `hours`:
* First 2 hours: Flat $5.

* Next 3 hours: $2/hr.

* After 5 hours: $1/hr.

* **Daily Max:** If hours >= 24, flat fee $20.


````python
def parking_fee(hours):
    if hours >= 24:
        return 20.0
    
    fee = 0
    if hours <= 2:
        fee = 5.0
    elif hours <= 5:
        fee = 5.0 + (hours - 2) * 2.0
    else:
        fee = 5.0 + 6.0 + (hours - 5) * 1.0
        
    return fee

# result:
f"The Parking fee for the Vehicle:",parking_fee(1)
````

### Q3. Password Strength Meter [Easy] (2 Marks)
**Scenario:**
Return "Weak" (<6 chars), "Medium" (only letters OR only numbers), or "Strong" (both letters AND numbers).


````python
def pass_word(password):
    if len(password) < 6:
        return "The Password is Weak"
    
    letters = any(c.isalpha() for c in password)
    numbers = any(c.isdigit() for c in password)
    
    if letters and numbers:
        return "The Password is Strong"
    else:
        return "The Password is Medium"
pass_word('Hello@2026')
````

### Q6. Anagram Checker [Easy] (2 Marks)
**Scenario:**
Check if two strings are anagrams (contain same letters), ignoring case.


````python
def is_anagram(str1, str2):
    # Normalize by converting to lowercase and removing spaces
    s1 = sorted(str1.lower().replace(" ", ""))
    s2 = sorted(str2.lower().replace(" ", ""))
    return s1 == s2

#Inputs
str1=input('Enter the Text1:')
str2=input('Enter the Text2:')
# Result
is_anagram(str1, str2)
````

### Q7. Character Frequency [Medium] (3 Marks)
**Scenario:**
Count frequency of characters in a string and return a dictionary.


````python
def char_frequency(str1):
    freq = {}
    for char in str1:
        freq[char] = freq.get(char, 0) + 1
    return freq

#Input
str1= input('Enter the Text:')
# Result
print(char_frequency(str1))
````

### Q8. Grouping by Category [Medium] (4 Marks)
**Scenario:**
Transform a list of dicts `[{'name': 'A', 'category': 'X'}, ...]` into `{'X': ['A'], ...}`.


````python
data = [{'name': 'A', 'category': 'X'},
    {'name': 'B', 'category': 'Y'},
    {'name': 'C', 'category': 'X'}]

grouping = {}
for item in data:
    category = item['category']
    name = item['name']
    grouping.setdefault(category, []).append(name)
    
print(grouping)
````

### Q9. Top Performer Finder [Medium] (3 Marks)
**Scenario:**
Given `{'Name': [scores]}`, find the student with the highest *average* score.


````python
# Sample data
student_scores = {'Bob': [85, 90, 78],
    'Marlie': [92, 88, 95],
    'yesha': [70, 80, 75]}
highest_avg = -1
top_student = None

for student, scores in student_scores.items():
    current_avg = sum(scores) / len(scores)
    
    if current_avg > highest_avg:
        highest_avg = current_avg
        top_student = student
print(f"Top Performer is {top_student} with an average of {highest_avg}")
````

### Q11. Flatten Nested List [Hard] (5 Marks)
**Scenario:**
Flatten `[1, [2, 3], [4, [5, 6]], 7]` using recursion.


````python
def flatten(nested_list):
    flat_list = []
    for item in nested_list:
        if isinstance(item, list):
            flat_list.extend(flatten(item))
        else:
            flat_list.append(item)
    return flat_list

input = [1, [2, 3], [4, [5, 6]], 7]
result = flatten(input)
result
````

You are given a NumPy array representing daily transaction amounts (in ₹):

a) Calculate the mean and standard deviation of the transaction amounts.

```python
transactions = np.array([

    9000, 3800, 2600, 6700, 4800

````python
import numpy as np

# Define the transactions array
transactions = np.array([1200, 4500, 3200, 8000, 1500,
                        6200, 7100, 2900, 5400, 4100,
                        9000, 3800, 2600, 6700, 4800])
#a) Calculate the mean and standard deviation of the transaction amounts.
mean_value=np.mean(transactions)
standard_deviation= np.std(transactions)
print(70*'-',f'\nStandard Deviation: {standard_deviation}\n Mean Value: {mean_value}')
#b) Identify all transaction values greater than the mean.
greater_mean = transactions[transactions > mean_value]
print(70*'-','\nThe value greater than Mean is: ',greater_mean)
#c) Replace all transaction values less than ₹3000 with ₹3000.
copied_transactions = transactions.copy()
copied_transactions[copied_transactions < 3000] = 3000
print(70*'-','\nAfter replacing values less than ₹3000 with ₹3000:\n',copied_transactions)
````

### Q13. Data Visualization & Interpretation – Revenue Analysis [Hard] (4 Marks)

You are given a dataset **`sales.csv`** with the following columns:

- `OrderDate` – date of purchase  
- `Region` – customer region  
- `Revenue` – revenue generated from the order  

---

#### **Tasks**

1. Load the dataset using **pandas** and convert `OrderDate` into datetime format.  
2. Aggregate the data to calculate:
   - Total revenue **per month**
   - Total revenue **per region**
3. Create the following visualizations:
   - A **line plot** showing the monthly revenue trend.
   - A **bar chart** comparing total revenue across regions.
4. Write **2–3 clear observations** interpreting what the plots indicate about revenue trends and regional performance.

---

####  **Constraints**

- Use **Matplotlib or Seaborn** for visualizations.  
- Do not hard-code any values.  
- Handle missing revenue values appropriately.



````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
# to suppress the warnings..
warnings.filterwarnings('ignore', category=FutureWarning)

#1. Load the dataset using pandas and convert OrderDate into datetime format.#Read the data from the CSV file
df = pd.read_csv(r"C:\Users\suman\Downloads\Python Exam\sales - LMS (2).csv")
#check if there is any Null Values..
df.isnull().sum()
#As there is no null values  in the dataframe no need for the fill and drop
# there is a null data in the revenue variable
df = df.dropna(subset=['Revenue']) 
#Extracting the time components from the orderdate variable..
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

#2. Aggregate the data to calculate:
#   - Total revenue **per month**
#   - Total revenue **per region**

monthly_revenue = df.set_index('OrderDate').resample('M')['Revenue'].sum().reset_index()
regional_revenue = df.groupby('Region')['Revenue'].sum().reset_index()
print('Total revenue per month\n',monthly_revenue)
print('\nTotal revenue per Region\n',regional_revenue)

#3. Create the following visualizations:
 #  - A **line plot** showing the monthly revenue trend.
  # - A **bar chart** comparing total revenue across regions.

plt.figure(figsize=(14, 6))
# Plot 1: Monthly Revenue Trend (Line Plot)
plt.subplot(1, 2, 1)
sns.lineplot(data=monthly_revenue, x='OrderDate', y='Revenue', marker='o')
plt.title('Monthly Revenue Trend')
plt.xticks(rotation=45)
# Plot 2: Total Revenue by Region (Bar Chart)
plt.subplot(1, 2, 2)
sns.barplot(data=regional_revenue, x='Region', y='Revenue', palette='viridis')
plt.title('Total Revenue Across Regions')

plt.tight_layout()
plt.show()

#Observations
observations = """ 
The East region is the top performance market & it generates the highest total revenue of 1,036,288.
The North region follows closely as the second-largest market at 876,438.
The graph represents nearly half of the sales driven by the leading East region.
The business achieved its maximum revenue in February at 415,669 & secondary peak occurred in September at 364,225.
The lowest performance period was recorded in August at 192,578.
"""
````

### Q14. Student Performance Analysis [Medium] (3 Marks)


You are given a CSV file **`students.csv`** with the following columns:

- `StudentID` – unique student identifier  
- `Subject` – subject name  
- `Marks` – marks obtained (out of 100)  
- `Attendance` – attendance percentage  

### **Tasks**

1. Load the dataset using **pandas**.
2. Calculate the **average marks for each subject**.
3. Find the **number of students** who have:
   - Marks greater than **60**
   - Attendance greater than or equal to **75%**
4. Add a new column `Result`:
   - `"Pass"` if Marks ≥ 40 **and** Attendance ≥ 75  
   - `"Fail"` otherwise

---

### **Constraints**

- Use **pandas only** (no NumPy, no loops).
- Use vectorized operations.
- Print all results clearly.

````python
import pandas as pd

# 1. Load the dataset using pandas.
#read the data set from the CSV file
df = pd.read_csv(r"C:\Users\suman\Downloads\Python Exam\students - LMS.csv")

# 2. Calculate average marks for each subject
avg_marks_per_subject = df.groupby('Subject')['Marks'].mean()
print("Average Marks per Subject:\n",avg_marks_per_subject)
print("-" * 60)

# 3. Find the number of students meeting specific criteria
marks_gt_60 = df[df['Marks'] > 60].shape[0]
attendance_ge_75 = df[df['Attendance'] >= 75].shape[0]

print(f"Number of students with Marks greater than 60: {marks_gt_60}")
print(f"Number of students with Attendance greater than 75%: {attendance_ge_75}")
print("-" * 60)

# 4. Add a new column 'Result' using vectorized operations
df['Result'] = 'Fail'
df.loc[(df['Marks'] >= 40) & (df['Attendance'] >= 75), 'Result'] = 'Pass'

print("Updated Dataframe with Results:")
df
````

### Q15. Log File Analyzer [Medium] (4 Marks)
**Scenario:**
Create a dummy log file, read it, and count occurrences of "ERROR".


````python
# 1. Dummy log file
log_content = """INFO: System started
ERROR: Database connection failed
INFO: Retrying...
ERROR: Database connection failed again
INFO: Admin logged in
ERROR: Unauthorized access attempt
INFO: Retrying...
ERROR: Database Error
INFO: Admin logged in
ERROR: Unauthorized access
"""
#'w' is for write the file and named it as app.log
with open("app.log", "w") as file:
    file.write(log_content)
error_count = 0
#'r' is for reading the file app.log
with open("app.log", "r") as file:
    for line in file:
        if "ERROR" in line:
            error_count += 1

print(f"Total occurrences of 'ERROR': {error_count}")
````
