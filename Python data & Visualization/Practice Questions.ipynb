```python
#Q7. 
def count_not_present(user_str, ref_string):
    # Base case: if string is empty
    if not user_str:
        return 0
    char = user_str[0]
    # Check if char is a space or exists in ref_string
    count = 0
    if char != " " and char not in ref_string:
        count = 1
# Recursive call for the rest of the string
    return count + count_not_present(user_str[1:], ref_string)

# Input and Setup
my_string = 'Data Science'
user_input = input("Enter a string: ")

result = count_not_present(user_input, my_string)
print(f"Count of elements not in '{my_string}': {result}")
```

```python
#Q9.
def print_pascal_triangle(n):
    row = [1]
    for i in range(n):
        # Print the current row
        print(row)
        
        # Calculate the next row based on the current row
        next_row = [1]
        for j in range(len(row) - 1):
            next_row.append(row[j] + row[j+1])
        next_row.append(1)
        
        row = next_row

# Print Pascal's Triangle up to 6 steps
print_pascal_triangle(6)
```

```python
#11. Sort a list by the first letter (Manual Sort)
# Sample list
items = ["Orange", "apple", "Banana", "cherry"]

n = len(items)
for i in range(n):
    for j in range(0, n - i - 1):
        # Compare first letters in lowercase
        if items[j][0].lower() > items[j + 1][0].lower():
            # Swap elements
            items[j], items[j + 1] = items[j + 1], items[j]

print("Sorted list:", items)
```

```python
#14.
# Initialize matrices
X = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]

Y = [[1, 8, 1],
     [3, 7, 3],
     [4, 5, 9]]

# Result matrix initialized with zeros
result = [[0, 0, 0],
          [0, 0, 0],
          [0, 0, 0]]

# Perform multiplication
for i in range(len(X)):
    for j in range(len(Y[0])):
        for k in range(len(Y)):
            result[i][j] += X[i][k] * Y[k][j]

# Display result
for r in result:
    print(r)
```

```python
#13.
# Given data lists
positive = ['good', 'awesome', 'best', 'nice']
negative = ['worst', 'awful', 'bad']

tweets = [
    'This government policies are good', 
    'bad implementation', 
    'The way he played showed that he is one of the best players in the world', 
    'Her acting in the play was awesome', 
    'The wine tastes awful', 
    "It's nice to hear this little kid's laugh"
]

# Lists to store the classified comments
positive_comments = []
negative_comments = []

# Loop through each tweet to classify it
for tweet in tweets:
    # Convert tweet to lowercase for case-insensitive matching
    tweet_lower = tweet.lower()
    
    # Check for presence of positive keywords
    if any(word in tweet_lower for word in positive):
        positive_comments.append(tweet)
    # Check for presence of negative keywords
    elif any(word in tweet_lower for word in negative):
        negative_comments.append(tweet)

# Display the separated comments
print("Positive Comments:")
for comment in positive_comments:
    print(f"- {comment}")

print("\nNegative Comments:")
for comment in negative_comments:
    print(f"- {comment}")
```
