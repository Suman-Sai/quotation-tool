```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_insurance=pd.read_csv(r"D:\Backup 14.03.2021\Desktop\Imarticus Post\Online PGA 38\Python Prog\insurance_data_with_day.csv")

sns.scatterplot(x="bmi", y="claim", data=df_insurance)
plt.title("Relationship between BMI and Claim")
plt.xlabel("BMI")
plt.ylabel("Claim")
plt.show()
```
