#### Question 1: Central Tendency
A dataset contains the following values:
Dataset = [12, 15, 18, 20, 22, 25, 28, 30, 35, 100]
 
a) Calculate the Mean, Median, and Mode for this dataset.
b) Which measure of central tendency would you recommend to use for this dataset and why?


```python
#importing necessary libraries
import numpy as np
from collections import Counter

# The dataset
dataset = [12, 15, 18, 20, 22, 25, 28, 30, 35, 100]

# a) Calculating Mean, Median, and Mode
mean_val = np.mean(dataset)
median_val = np.median(dataset)

frequency = Counter(dataset)
max_freq = max(frequency.values())
modes = [k for k, v in frequency.items() if v == max_freq]
# displaying the results
print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
if len(modes) == len(dataset):
    print("Mode: No unique mode (all values occur with equal frequency)")
else:
    print(f"Mode: {modes}")
```

###### The Median ($23.5$) is the recommended measure of central tendency to use for this dataset.Reasoning: The dataset contains a severe upper outlier ($100$), while the remaining values are clustered tightly between $12$ and $35$. The mean is highly sensitive to extreme values and is pulled upwards to $30.5$, which is higher than $80\%$ of the data points in the set. The median is a robust measure that resists the influence of extreme outliers, providing a much more accurate reflection of the center of this skewed dataset.

#### Question 2: Measures of Dispersion
Given two datasets:
Dataset A: [5, 10, 15, 20, 25] &
Dataset B: [10, 12, 15, 18, 20]
a) Calculate the Range, Variance, and Standard Deviation for both datasets.
b) Which dataset has more variability? Justify your answer using statistical measures.

```python
import numpy as np

# The datasets
dataset_A = [5, 10, 15, 20, 25]
dataset_B = [10, 12, 15, 18, 20]

def analyze_dispersion(data, name):
    # Calculating Range
    data_range = np.max(data) - np.min(data)
    
    # Calculate Sample Variance and Standard Deviation
    sample_var = np.var(data, ddof=1)
    sample_std = np.std(data, ddof=1)
    
    # Calculate Population Variance and Standard Deviation
    pop_var = np.var(data, ddof=0)
    pop_std = np.std(data, ddof=0)
    
    print(f"=== {name} ===")
    print(f"Range: {data_range}")
    print(f"Sample Variance: {sample_var:.2f}")
    print(f"Sample Standard Deviation: {sample_std:.2f}")
    print(f"Population Variance: {pop_var:.2f}")
    print(f"Population Standard Deviation: {pop_std:.2f}\n")

# Run analysis for both datasets
analyze_dispersion(dataset_A, "Dataset A")
analyze_dispersion(dataset_B, "Dataset B")
```

###### Dataset A has significantly more variability than Dataset B. The Range spread from the maximum to the minimum value is twice as large in Dataset A ($20$) compared to Dataset B ($10$). The standard deviation of Dataset A ($\approx 7.91$ or $7.07$) is roughly double that of Dataset B ($\approx 4.12$ or $3.69$). This shows that the individual observations in Dataset A deviate much further from the center ($15$) than the observations in Dataset B.

#### Question 3: Correlation and Covariance
 The following data shows the relationship between Hours Studied (X) and Test Scores (Y):
 

 
a) Calculate the Covariance between Hours Studied and Test Scores.
b) Calculate the Correlation Coefficient (Pearson's r).
c) Interpret the relationship. Is it strong or weak?

 #### Question 4: Probability Concepts
 A deck contains 52 playing cards (26 red and 26 black, with 4 suits of 13 each).
a) What is the probability of drawing a red card OR a face card (J, Q, K)?
💡 Use the Addition Rule. Count carefully—some face cards are red.
b) If we draw two cards without replacement, what is the probability that both are aces?
c) Given that the first card drawn is a King, what is the probability the second card is also a King?

```python
#importing necessary library
from fractions import Fraction

# Total cards in a standard deck
total_cards = 52

# a) Probability of drawing a red card OR a face card
# Total red cards = 26 (Hearts and Diamonds)
# Total face cards = 12 (J, Q, K across 4 suits)
# Red face cards = 6 (J, Q, K of Hearts and Diamonds)
p_red = Fraction(26, total_cards)
p_face = Fraction(12, total_cards)
p_red_and_face = Fraction(6, total_cards)

# Using the Addition Rule: P(A OR B) = P(A) + P(B) - P(A AND B)
prob_a = p_red + p_face - p_red_and_face


#b) Probability that both cards are aces (without replacement)
# First card being an ace: 4 aces out of 52 cards
# Second card being an ace: 3 aces remaining out of 51 cards
prob_b = Fraction(4, total_cards) * Fraction(3, total_cards - 1)


#c) Probability that the second card is a King given the first card is a King
# Since 1 King is already drawn, there are 3 Kings left out of 51 remaining cards
prob_c = Fraction(3, total_cards - 1)


# Print the results
print("=== Result ===")
print(f"a) P(Red OR Face): {prob_a} ({float(prob_a):.4f} or {float(prob_a)*100:.2f}%)")
print(f"b) P(Both Aces): {prob_b} ({float(prob_b):.4f} or {float(prob_b)*100:.2f}%)")
print(f"c) P(Second King | First King): {prob_c} ({float(prob_c):.4f} or {float(prob_c)*100:.2f}%)")
```

#### Question 5: Normal Distribution and Z‑Scores
Student test scores are normally distributed with Mean (μ) = 75 and Standard Deviation (σ) = 10.
a) Calculate the Z-score for a student who scored 85.
b) What percentage of students scored between 65 and 85? (Use the 68–95–99.7 rule)
c) If a student wants to be in the top 5% of the class, what minimum score is needed?
💡 Top 5% means Z ≈ 1.645.

```python
# Given parameters
mean = 75
std_dev = 10

print("=== Normal Distribution and Z-Scores ===")

# --- a) Calculate the Z-score for a student who scored 85 ---
score_a = 85
z_score_a = (score_a - mean) / std_dev
print(f"a) Z-score for a score of {score_a}: {z_score_a:.1f}")


# --- b) Percentage of students between 65 and 85 (68-95-99.7 rule) ---
percentage_b = 68.0
print(f"b) Percentage of students between 65 and 85: {percentage_b}%")


# --- c) Minimum score needed to be in the top 5% (Z ≈ 1.645) ---
z_c = 1.645
score_c = mean + (z_c * std_dev)

print(f"c) Minimum score needed using given Z (1.645): {score_c:.2f}")
```

#### Question 6: Binomial Distribution
 A quality control team finds that 95% of products pass inspection. If they test 10 products:
a) What is the probability that exactly 8 products pass inspection?
b) What is the probability that at least 9 products pass inspection?
c) What is the expected number of products that will pass inspection?

```python
import scipy.stats as stats

# Given parameters for the Binomial Distribution
n = 10      # total number of products tested
p = 0.95    # probability of a product passing inspection

print("=== Binomial Distribution ===")

# --- a) Probability that exactly 8 products pass inspection ---
# Using Probability Mass Function (PMF)
prob_a = stats.binom.pmf(8, n, p)
print(f"a) P(Exactly 8 products pass): {prob_a*100:.2f}%")


# --- b) Probability that at least 9 products pass inspection ---
# "At least 9" means exactly 9 OR exactly 10 products pass
prob_b = stats.binom.pmf(9, n, p) + stats.binom.pmf(10, n, p)

print(f"b) P(At least 9 products pass): {prob_b*100:.2f}%")


# --- c) Expected number of products that will pass inspection ---
# Expected value (Mean) of a binomial distribution is n * p
expected_c = n * p
print(f"c) Expected number of passing products: {expected_c:.1f}")
```

#### Question 7: Hypothesis Testing Fundamentals
 A manufacturer claims that the average battery life is 50 hours.
A quality team tests 30 batteries and finds an average of 48 hours with a standard deviation of 5 hours.
a) State the Null Hypothesis (H₀) and Alternative Hypothesis (Hₐ) for a two‑tailed test.
b) What is the test statistic (Z or t‑score)? (Use t‑test since sample size = 30)
c) If the p‑value is 0.042 and α = 0.05, should we reject or fail to reject H₀? What does this mean in practical terms?

```python
import numpy as np

# Given parameters
mu_0 = 50       # Hypothesized population mean
n = 30          # Sample size
x_bar = 48      # Sample mean
s = 5           # Sample standard deviation

print("=== Hypothesis Testing ===")

# --- b) Calculate the t-test statistic ---
standard_error = s / np.sqrt(n)
t_statistic = (x_bar - mu_0) / standard_error

print(f"b) Calculated t-statistic: {t_statistic:.4f}")

# --- c) Decision making based on given p-value ---
given_p_value = 0.042
alpha = 0.05

print(f"c) Decision Rule Evaluation:")
if given_p_value <= alpha:
    print(f"   Since p-value ({given_p_value}) <= alpha ({alpha}), we REJECT the null hypothesis (H0).")
else:
    print(f"   Since p-value ({given_p_value}) > alpha ({alpha}), we FAIL TO REJECT the null hypothesis (H0).")

#conclusion: In practical terms, there is statistically significant evidence to conclude that the manufacturer's claim is incorrect. The true average life of the batteries is significantly different (specifically, shorter) than $50$ hours.
```

#### Question 8: Confidence Intervals
 A sample of 50 customers has a mean spending of $120 with a standard deviation of $25.
a) Calculate a 95% confidence interval for the population mean spending.
💡 Use: CI = Mean ± (1.96 × SE), where SE = SD / √n.
b) Interpret what this confidence interval means.
c) If the company wants a 99% confidence interval, what would the new interval be?
💡 Use 2.576 for 99% confidence.

```python
import numpy as np

# Given parameters
n = 50          # Sample size
mean = 120      # Sample mean spending ($)
sd = 25         # Sample standard deviation ($)

print("=== Confidence Intervals ===")

# Calculate Standard Error (SE)
se = sd / np.sqrt(n)
print(f"Standard Error (SE): {se:.4f}")

# --- a) Calculate a 95% confidence interval ---
z_95 = 1.96
margin_error_95 = z_95 * se
ci_95_lower = mean - margin_error_95
ci_95_upper = mean + margin_error_95

print(f"a) 95% Confidence Interval: [${ci_95_lower:.2f}, ${ci_95_upper:.2f}]")
# --- b) Interpretation of confidence interval ---
print(f"b) Interpretation: We are 95% confident that the true population mean spending lies within the interval [${ci_95_lower:.2f},$ {ci_95_upper:.2f}].")

# --- c) Calculate a 99% confidence interval ---
z_99 = 2.576
margin_error_99 = z_99 * se
ci_99_lower = mean - margin_error_99
ci_99_upper = mean + margin_error_99

print(f"c) 99% Confidence Interval: [${ci_99_lower:.2f}, ${ci_99_upper:.2f}]")
```

#### Question 9: Independent Samples T‑Test
 A marketing team wants to compare customer satisfaction between two store locations.
Store A (n=15): Mean Satisfaction = 7.2, SD = 1.1
Store B (n=18): Mean Satisfaction = 6.8, SD = 1.4
a) State the null and alternative hypotheses.
b) Calculate the t‑statistic for this independent sample t‑test.
💡 t = (Mean₁ − Mean₂) / √[(SD₁²/n₁) + (SD₂²/n₂)]
c) With df ≈ 31 and α = 0.05, the critical value is approximately ±2.042. Is this difference statistically significant?
d) Interpret your findings. What can we conclude about satisfaction differences between stores?

```python
import numpy as np

# Given parameters
n1, mean1, sd1 = 15, 7.2, 1.1  # Store A
n2, mean2, sd2 = 18, 6.8, 1.4  # Store B

# --- a) Null and Alternative hypotheses ---
#Null Hypothesis (H0): There is no difference in the true mean customer satisfaction between Store A and Store B.
#Alternative Hypothesis (H1): There is a difference in the true mean customer satisfaction between Store A and Store B.

print("=== Independent Samples T-Test ===")

# --- b) Calculate the t-statistic ---
# Formula provided: t = (Mean1 - Mean2) / sqrt((SD1^2 / n1) + (SD2^2 / n2))
numerator = mean1 - mean2
denominator = np.sqrt((sd1**2 / n1) + (sd2**2 / n2))
t_statistic = numerator / denominator

print(f"b) Calculated t-statistic: {t_statistic:.4f}")

# --- c) Evaluation and Decision ---
critical_value = 2.042  # Given for df = 31, alpha = 0.05
alpha = 0.05

print(f"\nc) Decision Rule Evaluation:")
print(f"   Absolute t-statistic (|{t_statistic:.4f}|) vs Critical Value ({critical_value})")

# --- d) Conclusion based on the evaluation ---
print(f"\nd) Conclusion and Interpretation:")
if abs(t_statistic) >= critical_value:
    print("   Since |t| >= critical value, the difference IS statistically significant. (Reject H0)")
else:
    print("   Since |t| < critical value, the difference IS NOT statistically significant. (Fail to reject H0)")

#conclusion : There is insufficient statistical evidence to conclude that a true difference in customer satisfaction exists between Store A and Store B.
```

#### Question 10: One‑Way ANOVA
 A company tests employee productivity across 4 different work schedules:
Schedule A: n=12, Mean Productivity = 78, SD = 8
Schedule B: n=12, Mean Productivity = 82, SD = 7
Schedule C: n=12, Mean Productivity = 85, SD = 9
Schedule D: n=12, Mean Productivity = 80, SD = 8
a) What is the purpose of using ANOVA instead of multiple pairwise t‑tests?
b) Why would multiple t‑tests increase the chance of Type I error?
c) If the F‑statistic is 4.25 with p‑value = 0.008 (α = 0.05), what is your conclusion?
d) What is the next step after finding a significant ANOVA result?

###### Part a) The main purpose of using an Analysis of Variance (ANOVA) is to compare the means of three or more groups simultaneously using a single omnibus test. Instead of performing multiple separate tests between every possible pair of groups, ANOVA tests the global null hypothesis to determine if at least one group mean is significantly different from the others.

###### Part b) Every time you perform a statistical hypothesis test, there is a built-in probability of committing a Type I error (falsely rejecting a true null hypothesis), which is determined by the significance level ($\alpha$, usually $0.05$ or $5\%$).
###### ANOVA controls this risk by keeping the overall Type I error rate strictly at your chosen alpha level ($\alpha = 0.05$).

```python
# Given parameters from the problem scenario (Part c)
f_statistic = 4.25
p_value = 0.008
alpha = 0.05

print("=== One-Way ANOVA ===")

# --- c) Conclusion based on the provided F-statistic and p-value ---
print("c) Hypothesis Test Conclusion:")
print(f"   Significance Level (alpha): {alpha}")
print(f"   Calculated F-statistic:     {f_statistic}")
print(f"   Calculated p-value:         {p_value}")

if p_value <= alpha:
    print(f"   Decision: REJECT the null hypothesis (H0).")
    print(f"   Conclusion: There is a statistically significant difference in employee ")
    print(f"               productivity across the 4 different work schedules.")
else:
    print(f"   Decision: FAIL TO REJECT the null hypothesis (H0).")
    print(f"   Conclusion: There is no statistically significant difference in employee ")
    print(f"               productivity across the work schedules.")

# --- d) Demonstrating the Next Step (Post-Hoc Analysis Example) ---
print("\nd) Next Step Demonstration:")
print("   Since the ANOVA result is significant, a Post-Hoc test ")
print("   must be conducted to identify exactly which pairs of schedules differ.")
```

####  Question 14: Central Limit Theorem

* The Maturation Effect
