```python
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor

# Create presentation and blank slide
prs = Presentation()
slide = prs.slides.add_slide(prs.slide_layouts[6])

# Add a solid background (faster than gradient)
background = slide.background
fill = background.fill
fill.solid()
fill.fore_color.rgb = RGBColor(30, 144, 255)  # Dodger Blue

# Add title text
title = slide.shapes.add_textbox(Inches(1.5), Inches(2), Inches(7), Inches(1))
title_frame = title.text_frame
title_frame.text = "Weather Forecast Dashboard"
title_frame.paragraphs[0].font.size = Pt(44)
title_frame.paragraphs[0].font.bold = True
title_frame.paragraphs[0].font.color.rgb = RGBColor(255, 255, 255)

# Add subtitle
subtitle = slide.shapes.add_textbox(Inches(1.5), Inches(3), Inches(7), Inches(1))
subtitle_frame = subtitle.text_frame
subtitle_frame.text = "Powered by Power BI & WeatherAPI"
subtitle_frame.paragraphs[0].font.size = Pt(24)
subtitle_frame.paragraphs[0].font.color.rgb = RGBColor(255, 255, 255)

# Save directly to Desktop
prs.save(r"C:\Users\suman\OneDrive\Desktop\Weather_Forecast_Theme.pptx")
```
