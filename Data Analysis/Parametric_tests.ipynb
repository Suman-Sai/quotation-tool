###    Z-Test

```python
import numpy as np
import scipy.stats as stats

# 1. Define the given parameters
mu_0 = 10.0      # Claimed population mean
sigma = 0.5      # Known population standard deviation
n = 50           # Sample size
x_bar = 10.1     # Sample mean
alpha = 0.05     # Significance level

# 2. Calculate the Z-statistic
se = sigma / np.sqrt(n)  # Standard error
z_stat = (x_bar - mu_0) / se

# 3. Calculate critical values for a two-tailed test
z_critical = stats.norm.ppf(1 - alpha / 2)

# 4. Calculate the p-value
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))

# 5. Output results
print(f"Calculated Z-statistic: {z_stat:.2f}")
print(f"Critical Z-values:     ±{z_critical:.2f}")
print(f"P-value:                {p_value:.4f}")

# Decision logic
if abs(z_stat) > z_critical:
    print("Decision: Reject the null hypothesis (H0).")
else:
    print("Decision: Fail to reject the null hypothesis (H0).")
```

```python
#H0: The mean strength of the material is 55 lbs (μ = 55)
#H1: The mean strength of the material is not 55 lbs (μ ≠ 55)   
import pandas as pd 
from statsmodels.stats import weightstats as stests
df = pd.read_csv(r'C:\Users\suman\OneDrive\Documents\Python Documents\Strength.csv')  
stests.ztest(df['Strength(in lbs)'], value=55)
#Conclusion: Since the p-value is less than the significance level (0.05), we reject the null hypothesis (H0) and conclude that there is sufficient evidence to suggest that the mean strength of the material is not 55 lbs.
#That means the material does not meet the required strength of 55 lbs and claim is false.
```

```python
#H0: The Average yield due to fertilizer1 is equal to the average yield due to fertilizer2 (μ1 = μ2)
#H1: The Average yield due to fertilizer1 is not equal to the average yield due to fertilizer2 (μ1 ≠ μ2)
import pandas as pd                             
from statsmodels.stats import weightstats as stests
df = pd.read_csv(r'C:\Users\suman\OneDrive\Documents\Python Documents\Fertilizer.csv')
stests.ztest(df['Fertilizer 1'], df['Fertilizer 2'], value=0)
#Conclusion: Since the p-value is greater than the significance level (0.05), we fail to reject the null hypothesis (H0) and conclude that there is insufficient evidence to suggest that the average yield due to fertilizer1 is equal to  the average yield due to fertilizer2.
#That means there is no significant difference in the average yield between the two fertilizers.
```
