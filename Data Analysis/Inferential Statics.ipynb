```python
import numpy as np
#%pip install statsmodels
from statsmodels.stats.weightstats import ztest
# one-sample z-test
data = np.array([12.1, 13.5, 11.8, 14.2, 12.9, 13.0, 12.7])
stat, p_value = ztest(data, value=12.0)
print("One-sample z-test stat:", stat)
print("One-sample p-value:", p_value)

# two-sample z-test
data1 = np.array([12.1, 13.5, 11.8, 14.2, 12.9])
data2 = np.array([11.0, 10.8, 11.5, 10.9, 11.2])
stat2, p_value2 = ztest(data1, data2)
print("Two-sample z-test stat:", stat2)
print("Two-sample p-value:", p_value2)
```
