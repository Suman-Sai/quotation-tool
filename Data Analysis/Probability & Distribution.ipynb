```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define parameters
# We will use an Exponential distribution (highly skewed, definitely not normal)
num_samples = 10000   # Number of sample means to calculate
sample_size = 50      # Size of each individual sample (n >= 30 is ideal for CLT)

# 2. Generate the sample means
sample_means = []
for _ in range(num_samples):
    # Draw a sample of size 'n' from an exponential distribution
    sample = np.random.exponential(scale=2.0, size=sample_size)
    # Calculate the mean of this specific sample
    sample_means.append(np.mean(sample))

# 3. Plot the result
plt.figure(figsize=(8, 5))
plt.hist(sample_means, bins=30, density=True, edgecolor='black', alpha=0.7, color='skyblue')

# Add labels and a title
plt.title(f'Central Limit Theorem (Sample Size n = {sample_size})', fontsize=14)
plt.xlabel('Sample Means', fontsize=12)
plt.ylabel('Density', fontsize=12)
plt.grid(axis='y', alpha=0.3)

# Show the classic bell curve shape
plt.show()
```
