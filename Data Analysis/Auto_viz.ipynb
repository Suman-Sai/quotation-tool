```python
import pandas as pd
from autoviz import AutoViz_Class
from autoviz import FixDQ 

# Initialize the AutoViz class
AV = AutoViz_Class()

%matplotlib inline

path = r"C:\Users\suman\Downloads\PythonDocuments\Superstore.csv"
df=pd.read_csv(path, encoding='latin1')

fixer = FixDQ()
fixer.fit(df)
df_clean = fixer.transform(df)
# Generate automated visualizations by passing your data file
# You can leave the first argument as an empty string if you are passing a dataframe instead
dft = AV.AutoViz(
    filename= "", 
    sep=",", 
    depVar="", 
    dfte=df_clean, 
    header=0, 
    verbose=2, # Changed from 1 to 2 to force plot displays
    lowess=False, 
    chart_format="inline", 
    max_rows_analyzed=150000, 
    max_cols_analyzed=30
)
```
