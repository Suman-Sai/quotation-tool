Power BI is a business intelligence tool developed by Microsoft that transforms raw data into interactive dashboards and meaningful insights

It allows you to connect to various data sources, clean and model the data, build insightful visualizations

Poer Bi isused for : 
* Import data from Excel, SQL, CSV, JSON, APIs and websites
* Clean and transform data without heavy coding
* Create interactive charts, KPIs, dashboards and reports
* Analyze trends using filters and slicers
* Share dashboards securely with teams
* Automate data refresh
