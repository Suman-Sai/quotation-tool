```python
from scipy.stats import chisquare
obs_freq=[280,190,90]
exp_freq=[0.43*560, 0.48*560, 0.09*560]

result = chisquare(obs_freq, exp_freq)
print(result)
#Since the p-val(0.00) is less than the los(0.05), it is strong evidence against the null hypothesis, so the null hypothesis is rejected. We may conclude
#that the Expected and Observerd frequncies are different
```
