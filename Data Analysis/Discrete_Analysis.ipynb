```python
import numpy as np
rain_mm= [21.2,21,98,25,22.8,22.1,20.8,24.7,24.33,22]
#Average rainfall in last 10days 
print('The Average rainfall in the last 10 days is: ',np.mean(rain_mm))
# Here 98 is the outlier which deviates the result mean from the actual average rainfall in the last 10 days so remove the outlier and calculate the data again.
print('The Average rainfall in the last 10 days is: ',np.median(rain_mm))
```

```python
import pandas as pd
# consider the series of daily salary of 15 workers
daily_sal = pd.Series([212, 121, 398, 245, 228, 275, 208,
                       212, 293, 275, 190, 514, 275, 322, 312])

# calculate the mode - here 0 is the index position and 275 appeared 3times in the data so its the mode value...!
print('Mode: ', daily_sal.mode())
```

```python
daily_sal = pd.Series([212, 121, 398, 245, 228, 275, 208,
                       212, 293, 275, 190, 514, 275, 322, 312])
#The First Quantileas Q1
print('Q1:' , daily_sal.quantile(0.25))
#the second quantile which is mean as Q2
print('Q2:' , daily_sal.quantile(0.5))
#The Third Quantileas Q3
print('Q3:' , daily_sal.quantile(0.75))
```

```python
#The First Deciles as D1
print('D1:' , daily_sal.quantile(0.10))
#The Second Deciles as D2
print('D2:' , daily_sal.quantile(0.20))
#The Third Deciles as D3
print('D3:' , daily_sal.quantile(0.30))
#The Fourth Deciles as D9
print('D9:' , daily_sal.quantile(0.90))
```

```python
#the first percentile as P1
print('P1:' , daily_sal.quantile(0.01))
#the second percentile as P2
print('P2:' , daily_sal.quantile(0.02))
#the third percentile as P3
print('P3:' , daily_sal.quantile(0.03))
#the ninety ninth percentile as P99
print('P99:' , daily_sal.quantile(0.99))
```

```python
import numpy as np

# consider the sales of shop_A for 10 days
sales_A = np.array([412, 329, 309, 325, 378, 321, 428, 319, 353, 482])

# obtain the range
# compute the difference between maximum and minimum value from the array
print("Range of sales_A:", sales_A.max() - sales_A.min())


# consider the sales of shop_B for 10 days
sales_B = np.array([452, 479, 309, 425, 478, 321, 438, 319, 453, 482])

# obtain the range
# compute the difference between maximum and minimum value from the array
print("Range of sales_B:", sales_B.max() - sales_B.min())
```

```python
# consider the series of daily salary of 15 workers
daily_sal = pd.Series([212, 121, 398, 245, 228, 275, 208,
                       212, 293, 275, 190, 514, 275, 322, 312])

# calculate first quartile
Q1 = daily_sal.quantile(0.25)

# calculate third quartile
Q3 = daily_sal.quantile(0.75)

# calculate the IQR
print('IQR:', Q3 - Q1)
```

```python
#Variance
# consider the rainfall for 10 days in mm
rain_mm = [41.2, 21.6, 58, 25, 32.8, 22.1, 30.8, 24.7, 34.33, 22]

# calculate the variance
print('Variance:', np.var(rain_mm))
#Standard Deviation
# consider the rainfall for 10 days in mm
print('Standard Deviation:', np.std(rain_mm))
#Standard Deviatio n is the square root of variance
print('Standard Deviation:', np.sqrt(np.var(rain_mm)))
```

```python
# --- Location A ---
# expected salary of employees at Location_A
expected_sal_A = 568

# standard deviation in salary at location_A
std_sal_A = 20

# calculate C.V.
cv = (std_sal_A/expected_sal_A)*100
print('C.V. for Location A:', cv)


# --- Location B ---
# expected salary of employees at Location_B
expected_sal_B = 780

# standard deviation in salary at Location_B
std_sal_B = 35

# calculate C.V.
cv = (std_sal_B/expected_sal_B)*100
print('C.V. for Location B:', cv)
```
