```python
records['created_date'] = pd.to_datetime(records['created_date'])
records['canceled_date'] = pd.to_datetime(records['canceled_date'])

signups = records.resample('M', on='created_date').size()
cancellations = records.resample('M', on='canceled_date').size()
plt.figure(figsize=(10, 6))
signups.plot(label='New Signups', color='green')
cancellations.plot(label='Cancellations', color='red')
plt.title('Signups vs. Cancellations Over Time')
plt.legend()
plt.show()
```
