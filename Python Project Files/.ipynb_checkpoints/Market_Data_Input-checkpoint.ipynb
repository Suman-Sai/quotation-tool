Big Market Data .....
    *The Sample Project is about reading the data from a text file and cleaning the data using pandas and adding the Observations.
    *Added the Profit percentage and profit or loss Column in to the data frame
    *Added the New Observations with the User Inputs.

```python
#Importing the Libraries
import pandas as pd
import numpy as np
#reading the .csv file and getting it into as Python data frames
records=pd.read_csv(r"C:\Users\suman\OneDrive\Documents\Python Documents\bigmarket.txt",sep='\t')
records
```

```python
#Duplicating the data (it wont affect the original data)
test=records.copy()
#getting the Information of the dataframe
print(test.info())
test.isnull().sum()
#if missing values is present fill the values with 0
test.fillna(0)
#to drop the Missing Values for temporary
test.dropna()
print("\nThe Overall data(Observations, Variables):",test.shape,"\nVariable Names:",test.columns)
```

```python
#renaming the Store as Branch 
test=test.rename(columns={'Store':'Branch'})
#Assigning the ProfitorLoss Column..!
test=test.assign(ProfitorLoss=np.where(test['Sales']>35000,'Profit','Loss'))
# Assigning the Profit or loss Percentage Column
test= test.assign(Profit=((test['Sales']*0.10)/test['Sales'])*100)
```

```python
#Which Month getting profits list...!
print('The Total Profits Count is: ',(test['ProfitorLoss']=='Profit').sum())
test.loc[test['ProfitorLoss'] == 'Profit',['Month','Branch','Sales','ProfitorLoss','Profit']]
```

```python
#Excluding the Branch A 
print((test['Branch']=='A').sum())
test[~(test['Branch']=='A')]
```

```python
#Query based conditions
test.query("Month in ['Jan','Feb'] and ProfitorLoss == 'Profit'")
test.query("Branch in ['A','C','D','E'] and Sales > 30000 and ProfitorLoss == 'Profit'")
```

```python
#Before Adding the Observations the information of the Data Frame...!
test.info()
```

```python
#Creating the New Observations with User Inputs...!
a=input('Enter the Month:')
b=input('Enter the Branch:')
c=int(input('Enter the Sale:'))
new_data = {
    'Month': a,
    'Branch': b,
    'Sales': c ,
    'ProfitorLoss': np.where(c>35000,'Profit','Loss'),
    'Profit': ((c*0.10)/c)*100
}
new_row=pd.DataFrame([new_data])
test=pd.concat([test,new_row], ignore_index=True)
test.tail()
```
