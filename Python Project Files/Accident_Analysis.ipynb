```python
#importing the pandas Library 
import pandas as pd
#importing the Numpy Library 
import numpy as np
#importing the Seaborn Library 
import seaborn as sns
#importing the MatPlot Library 
import matplotlib.pyplot as plt
#to avoid the warnings displaying in the execution part import warning library 
import warnings
# This will ignore the FutureWarnings globally for this cell
warnings.filterwarnings('ignore', category=FutureWarning)
```

```python
#Read the CSV file to extravt the data....
df_accident=pd.read_csv(r"C:\Users\suman\Downloads\RTA Dataset -  LMS.csv")

#Check the 5 random data using sample()
df_accident.sample(5)
```

```python
#check the dimensions of the data(observations and variables)
df_accident.shape
```

```python
#get the data information....
df_accident.info()
```

```python
#Check the Missing Values....
df_accident.isnull().sum()
```

```python
#Fill the Missing Values with 0 or any Values....
df_accident.fillna(0)
```

```python
#Converting the date and time using type cast...
df_accident.Time=pd.to_datetime(df_accident.Time, errors='coerce')
```

```python
#Recheck the data types....
df_accident.dtypes
```

```python
#extract the various time components()
df_accident=df_accident.assign(hour=df_accident.Time.dt.hour)
df_accident.head()
```

```python
#Drop the Columns using drop(), axis=1 drops the specific columns
df_accident=df_accident.drop(['day','Time'],axis=1)
```

```python
df_accident.info()
```

```python
# Generate the correlation matrix by explicitly ignoring strings
corr = df_accident.corr(numeric_only=True)

# Display the correlation matrix
corr
```

```python
#Generating the heat map...
# Set the figure size
plt.figure(figsize=(5,5))

# Create the heatmap
# annot=True adds the numbers into the squares
# cmap='coolwarm' makes positive correlations red and negative ones blue
sns.heatmap(corr, annot=True, cmap='RdYlGn', fmt=".2f", linewidths=0.5, annot_kws= {"size":10})

# Add a title
plt.title('Correlation Heatmap of Accident Data')
#set the text size using fontsize
plt.xticks(fontsize = 8)
plt.yticks(fontsize = 8)
plt.show()
```

```python
#Clear, evidence-based findings such as "Accidents peak between 5 PM and 7 PM" or "Rainy weather conditions significantly increase accident severity levels."

#Peak Accidents between timeframes plot....
df_accident['hour'].value_counts().sort_index().plot(title='Peak Accidents Time Frame',kind='line')
#Severity accident levels for weather severity plot...
#normalize= 'index' , will converts those counts into percentages (proportions) based on the row total.
#without normalize m it shows the raw data like daylight=50,000 accidents and all.....
pd.crosstab(df_accident['Light_conditions'], df_accident['Accident_severity'], normalize='index')
```

```python
#Professional charts showing accidents by time of day, severity by weather conditions, and vehicle type involvement patterns.

# --- Set Global Aesthetics ---
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 100

# Create a figure with three subplots
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# 1. ACCIDENTS BY TIME OF DAY (Line Chart)
# Grouping by hour and counting incidents
hourly_data = df_accident['hour'].value_counts().sort_index().reset_index()
hourly_data.columns = ['Hour', 'Count']

sns.lineplot(data=hourly_data, x='Hour', y='Count', marker='o', color='#2c3e50', ax=axes[0])
axes[0].fill_between(hourly_data['Hour'], hourly_data['Count'], color='#34495e', alpha=0.1)
axes[0].set_title('Accident Distribution by Hour', fontweight='bold')
axes[0].set_xticks(range(0, 24, 2)) # Simple way to show every 2nd hour

# 2. SEVERITY BY LIGHTING (Stacked Bar Chart)
# Using your cross-tab logic to show percentages
severity_pct = pd.crosstab(df_accident['Light_conditions'], 
                           df_accident['Accident_severity'], 
                           normalize='index') * 100

severity_pct.plot(kind='bar', stacked=True, ax=axes[1], color=['#e74c3c', '#f39c12', '#bdc3c7'])
axes[1].set_title('Severity % by Light Conditions', fontweight='bold')
axes[1].set_ylabel('Percentage (%)')
axes[1].legend(title='Severity', loc='center', bbox_to_anchor=(0.5,0.8), ncol=3)
axes[1].tick_params(axis='x', rotation=45)

# 3. VEHICLE TYPE INVOLVEMENT (Horizontal Bar Chart)
top_vehicles = df_accident['Type_of_vehicle'].value_counts().head(10)

sns.barplot(x=top_vehicles.values, y=top_vehicles.index, palette='viridis', ax=axes[2])
axes[2].set_title('Top 10 Involved Vehicle Types', fontweight='bold')
axes[2].set_xlabel('Total Incidents')

# Adjust layout to prevent label overlap
plt.tight_layout()
plt.show()
```
